```python
vectorizer = TfidfVectorizer(lowercase=True, 
                             max_features=100,
                             max_df=0.8,
                             min_df=5,
                            ngram_range= (1,3),
                            stop_words = "english"
                            )
```

```python
vectors = vectorizer.fit_transform(lines)
```

```python
#convert to a regular array
tf_idf_array = vectors.toarray()

#obtain the original terms in the corpus using get_feature_names

words_set = vectorizer.get_feature_names_out()
```

```python
features_names = vectorizer.get_feature_names_out()
```

```python
dense = vectors.todense()
```

```python
denselist = dense.tolist()
```

```python
len(denselist)
```

```python
all_keywords = []

for desc in denselist:
    x = 0
    keywords = []
    for word in desc:
        if word > 0:
            keywords.append(features_names)
            
        x += 1
    all_keywords.append(keywords)
```
