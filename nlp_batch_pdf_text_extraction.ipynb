```python
from tika import parser
```

```python
# PARSING THE CONTENTS OF THE PDF INTO TXT
# ENCODE THEM INTO UTF-8 FORMAT

# USE THE STRINGS STORED IN THE new_files list in naming the txt files

# WRITE THE output file into the "the_file"


# ".from_file" refers to the current working directory, that's why we had to
# change it as can be seen above.

count = 0
from os.path import dirname, abspath
from pathlib import Path

for file_name in os.listdir('C:/Users/USER/Desktop/Files/'):
    file_data = parser.from_file(file_name)
    
    output = file_data['content']
    
    output = output.encode('utf-8', errors= 'ignore')
    name = str(new_files[count])+ ".txt"
    
    with open(name,'w')  as the_file:
        the_file.write(str(output))
        
    count +=1
    
#     p = Path(folder_path).parent.joinpath(f"{cities}.txt")
#     if not p.is_file():
#         rest_df.to_csv(p,index=False)
#     else:
#         print('file exists')
        
#     count += 1
```
