```python
import nltk
from nltk.tokenize import sent_tokenize, sent_tokenize
nltk.download('vader_lexicon')
filename = "C:/Users/USER/Desktop/Strategie_Smart_City_Berlin_en.txt"
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
vds = SentimentIntensityAnalyzer()
```

```python
file = open(filename, "r", encoding= "utf-8")
text = file.read()
text = text.replace("\n"," ")
tokenizer = nltk.data.load("tokenizers/punkt/english.pickle")
sentences = sent_tokenize(text)
print (sentences)
```

```python
import pandas as pd

scores = []
for txt in sentences:
    vs = vds.polarity_scores(txt)
    scores.append(vs)
df = pd.DataFrame(sentences, columns= ['Text'])
df2 = pd.DataFrame(scores)
all_df = pd.concat([df,df2], axis=1)
```

```python
print(all_df)
```

```python
all_df.to_csv(r'C:/Users/USER/Desktop/VADER_results.csv', index=False)
```
